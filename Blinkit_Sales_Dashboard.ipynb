# SQL-Based Sales Analysis

This section uses SQL to analyze the Blinkit sales dataset and extract business insights. The analysis focuses on order volume, sales performance, product categories, outlet performance, customer ratings, and delivery efficiency.

### SQL Database Setup

Load the cleaned CSV into Pandas and create a SQLite database connection. The table `blinkit_sales` will be used for the SQL queries that follow.

```python
# Database setup
import pandas as pd
import sqlite3

df = pd.read_csv("blinkit_cleaned.csv")

conn = sqlite3.connect("blinkit.db")

df.to_sql(
    "blinkit_sales",
    conn,
    if_exists="replace",
    index=False
)

print("SQL database created successfully.")
print("Total records:", pd.read_sql_query(
    "SELECT COUNT(*) AS total_orders FROM blinkit_sales;",
    conn
).iloc[0, 0])
```

### Validate the SQL Table

Preview the first records from the database table to confirm that the data was loaded and the columns are accessible.

```python
# Validate the database
pd.read_sql_query("""
    SELECT *
    FROM blinkit_sales
    LIMIT 5;
""", conn)
```

### 1. Total Order Count

**Business Objective:** Determine the total number of orders recorded in the dataset.

**SQL Concept:** `COUNT()` aggregate function.



```python
query = """
SELECT COUNT(*) AS total_orders
FROM blinkit_sales;
"""

pd.read_sql_query(query, conn)
```

### 2. Total Sales and Average Order Value

**Business Objective:** Calculate the total sales value and average order value to understand the overall order-value performance.

**SQL Concepts:** `SUM()` and `AVG()` aggregate functions.

```python
query = """
SELECT
    ROUND(SUM(order_value), 2) AS total_sales,
    ROUND(AVG(order_value), 2) AS average_order_value
FROM blinkit_sales;
"""

pd.read_sql_query(query, conn)
```

### 3. Sales Analysis by Item Type

**Business Objective:** Identify which item types contribute the most to total sales.

**SQL Concepts:** `SUM()`, `GROUP BY`, and `ORDER BY`.

```python
query = """
SELECT
    item_type,
    SUM(order_value) AS total_sales,
    COUNT(*) AS total_orders
FROM blinkit_sales
GROUP BY item_type
ORDER BY total_sales DESC;
"""

pd.read_sql_query(query, conn)
```

### 4. Sales Analysis by City

**Business Objective:** Compare sales and order volume across cities to identify geographic differences in the current dataset.

**SQL Concepts:** `SUM()`, `COUNT()`, `GROUP BY`, and `ORDER BY`.



```python
query = """
SELECT
    city,
    SUM(order_value) AS total_sales,
    COUNT(*) AS total_orders,
    ROUND(AVG(order_value), 2) AS average_order_value
FROM blinkit_sales
GROUP BY city
ORDER BY total_sales DESC;
"""

pd.read_sql_query(query, conn)
```

### 5. Operational Performance Analysis

**Business Objective:** Evaluate outlet sales, delivery times, and customer ratings to identify operational patterns and potential data-quality issues.

```python
# 1. Outlet performance
outlet_query = """
SELECT outlet_id,
       SUM(order_value) AS total_sales,
       COUNT(*) AS total_orders
FROM blinkit_sales
GROUP BY outlet_id
ORDER BY total_sales DESC;
"""

# 2. Delivery performance
delivery_query = """
SELECT ROUND(AVG(delivery_minutes), 2) AS avg_delivery_value,
       MIN(delivery_minutes) AS min_delivery_value,
       MAX(delivery_minutes) AS max_delivery_value
FROM blinkit_sales;
"""

# 3. Rating summary
rating_query = """
SELECT ROUND(AVG(rating), 2) AS average_rating_value,
       MIN(rating) AS min_rating_value,
       MAX(rating) AS max_rating_value
FROM blinkit_sales;
"""

print("OUTLET PERFORMANCE")
display(pd.read_sql_query(outlet_query, conn))

print("DELIVERY SUMMARY")
display(pd.read_sql_query(delivery_query, conn))

print("RATING SUMMARY")
display(pd.read_sql_query(rating_query, conn))
```
